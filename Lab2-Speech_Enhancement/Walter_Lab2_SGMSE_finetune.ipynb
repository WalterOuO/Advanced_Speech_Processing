##4. Eval_reuslt

最後evaluation是用沒有finetune的版本  
* PESQ: 1.15 ± 0.14，算品質(比較嚴格)：只要有傷到signal就會很低，要做到3以上才
算好  
* ESTOI: 0.35 ± 0.19，比較寬鬆的指標，算聽不聽得懂他的聲音(會算能量的correlation)，對Noise比較robust(寬容)  
* SI-SDR: -1.6 ± 8.7  
* SI-SIR: 11.5 ± 16.4  
* SI-SAR: 0.8 ± 8.4  
  
SDR(speech-to-destortion ratio)：[ (s+n)-s ]/s 要 -> 0，speech 保的不好這就會很低，這結果來看超廢  
  
還有另一個指標沒用到的：  
SNR(speech denoise ratio)：e-n/s 要 -> 1


這結果超爛：因為混了很強的noise進去
  
評分Criterion: 五個 Mean 都要比上面高，Std可以一樣就好
