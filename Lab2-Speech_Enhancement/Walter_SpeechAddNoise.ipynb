```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Created on Thu Oct 24 16:22:51 2024

@author: peterwu
"""

import pandas as pd
import glob
import soundfile as sf
import numpy as np
import librosa
from random import sample

# VCTK_ROOT_train = '/content/try_set/train'
# VCTK_ROOT_eval = '/content/try_set/valid'
VCTK_ROOT_train = '/content/Dummy_set/train/clean'
VCTK_ROOT_eval = '/content/Dummy_set/valid/clean'
ESC50_ROOT = '/content/ESC-50-master'
# !rm /content/Dummy_set/train/noisy/*.wav
# !rm /content/Dummy_set/valid/noisy/*.wav
```

```python
def snr_noise_mix(s ,ns, snr):           # 用SNR_ratio混出音檔
    snr = 10**(snr/10.0)
    speech_power = np.sum(np.power(s,2))/len(s)
    noise_power = np.sum(np.power(ns,2))/len(ns)
    noise_update = ns / np.sqrt(snr * noise_power/speech_power)
    return noise_update + s

def s_n_p_align(s, ns):               # 對齊speech/noise，是改noise長度，跟speech對齊。沒有動到speech
    if len(s) > len(ns):
        new_ns = []
        count = len(s)//len(ns)+1
        for i in range(count):
            new_ns.append(ns)
        new_ns = np.hstack(new_ns)[:len(s)]
    elif len(s) <= len(ns):
        new_ns = ns[:len(s)]
    return s, new_ns
```

```python
speech_train_list = []
for file in glob.glob(VCTK_ROOT_train+'/*'):
  speech_train_list.append(file)
  #print(file)
speech_eval_list = []
for file in glob.glob(VCTK_ROOT_eval+'/*'):
  speech_eval_list.append(file)
```
