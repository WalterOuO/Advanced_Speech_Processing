```python
import string

def isEnglish(s):
    try:
        s.encode(encoding='utf-8').decode('ascii')
    except UnicodeDecodeError:
        return False
    else:
        return True

def check_punctuation(word):
    return any(char in string.punctuation for char in word)

def clean_txt(text_ori):    # 把text送入clean_text，可產生沒有亂碼與標點符號的乾淨text
    text_mdf = ''
    for t in text_ori:
        if isEnglish(t) or t == ' ':
            if not check_punctuation(t):
                text_mdf+=t.lower()
    return text_mdf
```

```python
# 用 Wiener Filter 降噪
import librosa
import soundfile as sf
import scipy.signal
import os
import numpy as np

# 定義資料夾路徑
input_dir = "Final_Target_Set/wav"  # 原始音檔資料夾
denoise_out_dir = "Final_Target_Set/wav_wiener"  # 降噪後的音檔輸出資料夾

# 確保輸出資料夾存在
os.makedirs(denoise_out_dir, exist_ok=True)
!rm /content/Final_Target_Set/wav_wiener/*

# 遍歷資料夾中的音檔
for filename in os.listdir(input_dir):
    if filename.endswith(".wav"):
        # 使用librosa加載音訊檔案
        file_path = os.path.join(input_dir, filename)
        y, sr = librosa.load(file_path, sr=None)  # sr=None 保持原始的取樣率

        # 檢查數據中是否有無效值
        if np.any(np.isnan(y)) or np.any(np.isinf(y)):
            print(f"檔案 {filename} 包含無效數值，跳過處理")
            continue  # 如果有nan或inf值，跳過該檔案

        # 可調的mySize大小
        mysize = 7     # 你可以嘗試不同的窗口大小
        noise = None
        gain = 1.5  # 增加音量
        denoised_data = scipy.signal.wiener(y, mysize=mysize, noise=noise)
        denoised_data = denoised_data * gain

        # 檢查降噪後的數據範圍
        print(f"{filename}: max = {np.max(denoised_data)}, min = {np.min(denoised_data)}")

        # 如果數據值仍然過大，可以進行縮放處理
        if np.max(np.abs(denoised_data)) > 1:
            denoised_data = denoised_data / np.max(np.abs(denoised_data))  # 確保值在[-1, 1]範圍內

        # 保存降噪後的音訊，使用soundfile.write保存
        output_path = os.path.join(denoise_out_dir, filename)
        sf.write(output_path, denoised_data, sr)

wavs_path = [path for path in glob.glob(denoise_out_dir+'/*.wav')]
```

```python
# noisereduce 降噪
import os
import librosa
import noisereduce as nr
import soundfile as sf

input_dir = "Final_Target_Set/wav"  # 原始音檔資料夾
denoise_outdir = "Final_Target_Set/wav_noisereduce"  # 降噪後的音檔輸出資料夾

os.makedirs(denoise_outdir, exist_ok=True)

for filename in os.listdir(input_dir):
    if filename.endswith(".wav"):
        file_path = os.path.join(input_dir, filename)
        print(f"處理音檔: {file_path}")

        y, sr = librosa.load(file_path, sr=None)  # 使用原始取樣率

        reduced_noise = nr.reduce_noise(y=y, sr=sr)

        output_path = os.path.join(denoise_outdir, filename)
        sf.write(output_path, reduced_noise, sr)
        print(f"已儲存降噪後的音檔至: {output_path}")

print("降噪處理完成！")

wavs_path = [path for path in glob.glob(denoise_outdir+'/*.wav')]
```

```python
# speechbrain 降噪
import os
from speechbrain.pretrained import SpectralMaskEnhancement
import torchaudio
import torch

# 設定資料夾路徑
input_dir = "Final_Target_Set/wav"  # 原始音檔資料夾
wav_metricgan = "Final_Target_Set/wav_metricgan"  # 降噪後的音檔輸出資料夾

# 確保輸出資料夾存在
os.makedirs(wav_metricgan, exist_ok=True)

# 載入 SpeechBrain 的預訓練模型 (MetricGAN+)
enhance_model = SpectralMaskEnhancement.from_hparams(
    source="speechbrain/metricgan-plus-voicebank",
    savedir="pretrained_models/metricgan-plus-voicebank"
)

# 遍歷所有音檔
for filename in os.listdir(input_dir):
    if filename.endswith(".wav"):
        file_path = os.path.join(input_dir, filename)
        print(f"處理音檔: {file_path}")

        # 讀取音檔
        noisy_signal, sr = torchaudio.load(file_path)
        signal_length = torch.tensor([noisy_signal.shape[1] / noisy_signal.shape[1]])

        # 使用模型進行降噪
        enhanced_signal = enhance_model.enhance_batch(noisy_signal, lengths=signal_length)

        # 儲存降噪後的音檔
        output_path = os.path.join(wav_metricgan, filename)
        torchaudio.save(output_path, enhanced_signal.squeeze(1), sr)
        print(f"已儲存降噪後的音檔: {output_path}")

print("降噪處理完成！")
```

## Eval WER

```python
import jiwer
import glob
# check align
ground_truths_root = 'Final_Target_Set/transcript/'
groundtruths_paths = sorted([path for path in glob.glob(ground_truths_root+'/*.txt')])
hypothesis_paths = sorted([path for path in glob.glob(decode_root+'/*.txt')])
for ele in groundtruths_paths:
  assert ele.split('/')[-1].split('.')[0] == hypothesis_paths[groundtruths_paths.index(ele)].split('/')[-1].split('.')[0]
# load decode result & GT
groundtruths = []
hypothesis = []
for path in groundtruths_paths:
  with open(path,'r') as f:
    groundtruths.append(f.read().replace('\n',''))
for path in hypothesis_paths:
  with open(path,'r') as f:
    hypothesis.append(f.read()[1:].replace('\n',''))
```

```python
wer_score = jiwer.wer(truth=groundtruths, hypothesis=hypothesis)
print(f"{wer_score=}")
```

```python
from jiwer import wer, RemovePunctuation, RemoveMultipleSpaces, Strip, ToLowerCase, Compose
transform = Compose([
    RemovePunctuation(),
    RemoveMultipleSpaces(),
    Strip(),
    ToLowerCase()
])

# 正規化處理
normalized_refs = [transform(ref) for ref in groundtruths]
normalized_res = [transform(re) for re in hypothesis]

# 計算 WER
error_rate = jiwer.wer(normalized_refs, normalized_res)

print(f"Normalized Refs: {normalized_refs}")
print(f"Normalized Res: {normalized_res}")
print(f"WER: {error_rate}")
```

```python
import re

def standardize_text(text):
    text = re.sub(r"(?<!\w)im(?!\w)", "i am", text)  # 'im' 替换为 'i am'
    text = re.sub(r"(?<!\w)theres(?!\w)", "there is", text)  # 'theres' 替换为 'there is'
    text = re.sub(r"(?<!\w)heis(?!\w)", "he is", text)  # 'heis' 替换为 'he is'
    text = re.sub(r"(?<!\w)shes(?!\w)", "she is", text)  # 'shes' 替换为 'she is'
    text = re.sub(r"(?<!\w)youre(?!\w)", "you are", text)  # 'youre' 替换为 'you are'
    text = re.sub(r"(?<!\w)ive(?!\w)", "I have", text)  # 'ive' 替换为 'I have'
    text = re.sub(r"(?<!\w)ill(?!\w)", "I will", text)  # 'ill' 替换为 'I will'
    text = re.sub(r"(?<!\w)wont(?!\w)", "will not", text)  # 'wont' 替换为 'will not'
    text = re.sub(r"(?<!\w)dont(?!\w)", "do not", text)  # 'dont' 替换为 'do not'
    text = re.sub(r"(?<!\w)cant(?!\w)", "cannot", text)  # 'cant' 替换为 'cannot'
    text = re.sub(r"(?<!\w)its(?!\w)", "it is", text)  # 'its' 替换为 'it is'

    # 其他常见缩写词替换
    text = re.sub(r"(?<!\w)gonna(?!\w)", "going to", text)  # 'gonna' 替换为 'going to'
    text = re.sub(r"(?<!\w)wanna(?!\w)", "want to", text)  # 'wanna' 替换为 'want to'
    text = re.sub(r"(?<!\w)kinda(?!\w)", "kind of", text)  # 'kinda' 替换为 'kind of'
    text = re.sub(r"(?<!\w)gotta(?!\w)", "got to", text)  # 'gotta' 替换为 'got to'
    text = re.sub(r"(?<!\w)lemme(?!\w)", "let me", text)  # 'lemme' 替换为 'let me'
    text = re.sub(r"(?<!\w)gimme(?!\w)", "give me", text)  # 'gimme' 替换为 'give me'
    text = re.sub(r"(?<!\w)outta(?!\w)", "out of", text)  # 'outta' 替换为 'out of'
    text = re.sub(r"(?<!\w)shoulda(?!\w)", "should have", text)  # 'shoulda' 替换为 'should have'
    text = re.sub(r"(?<!\w)woulda(?!\w)", "would have", text)  # 'woulda' 替换为 'would have'
    text = re.sub(r"(?<!\w)coulda(?!\w)", "could have", text)  # 'coulda' 替换为 'could have'
    text = re.sub(r"(?<!\w)ya(?!\w)", "you", text)  # 'ya' 替换为 'you'
    text = re.sub(r"(?<!\w)aint(?!\w)", "is not", text)  # 'aint' 替换为 'is not' 或根据上下文修改
    text = re.sub(r"(?<!\w)yall(?!\w)", "you all", text)  # 'yall' 替换为 'you all'
    text = re.sub(r"(?<!\w)howdya(?!\w)", "how do you", text)  # 'howdya' 替换为 'how do you'
    text = re.sub(r"(?<!\w)whatcha(?!\w)", "what are you", text)  # 'whatcha' 替换为 'what are you'
    text = re.sub(r"(?<!\w)whered(?!\w)", "where did", text)  # 'where'd' 替换为 'where did'
    text = re.sub(r"(?<!\w)whod(?!\w)", "who did", text)  # 'who'd' 替换为 'who did'
    text = re.sub(r"(?<!\w)isnt(?!\w)", "is not", text)  # 'isnt' 替换为 'is not'
    text = re.sub(r"(?<!\w)arent(?!\w)", "are not", text)  # 'arent' 替换为 'are not'
    text = re.sub(r"(?<!\w)wasnt(?!\w)", "was not", text)  # 'wasnt' 替换为 'was not'
    text = re.sub(r"(?<!\w)werent(?!\w)", "were not", text)  # 'werent' 替换为 'were not'
    text = re.sub(r"(?<!\w)havent(?!\w)", "have not", text)  # 'havent' 替换为 'have not'
    text = re.sub(r"(?<!\w)hasnt(?!\w)", "has not", text)  # 'hasnt' 替换为 'has not'
    text = re.sub(r"(laughter)+", "laughter", text)

    return text

# 对refs和res进行预处理时应用standardize_text
normalized1_refs = [transform(standardize_text(ref)) for ref in normalized_refs]
normalized1_res = [transform(standardize_text(re)) for re in normalized_res]

error_rate = jiwer.wer(normalized1_refs, normalized1_res)
print(f"WER: {error_rate}")
```
