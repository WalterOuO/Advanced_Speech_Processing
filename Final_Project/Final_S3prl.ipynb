## Decode

```python
import string

def isEnglish(s):
    try:
        s.encode(encoding='utf-8').decode('ascii')
    except UnicodeDecodeError:
        return False
    else:
        return True

def check_punctuation(word):
    return any(char in string.punctuation for char in word)

def clean_txt(text_ori):
    text_mdf = ''
    for t in text_ori:
        if isEnglish(t) or t == ' ':
            if not check_punctuation(t):
                text_mdf+=t.lower()
    return text_mdf
```

```python
import glob
import os
import torch
import librosa
import torch.nn.functional as F
from tqdm import tqdm
from s3prl.nn import S3PRLUpstream
from torchaudio.models.decoder import ctc_decoder

tokens = ["<blank>"] + list("abcdefghijklmnopqrstuvwxyz ")

decoder = ctc_decoder(
    lexicon=None,          # 沒有詞典時應設置為 None
    tokens=tokens,         # tokens 與 probs 的 vocab 尺寸應匹配
    beam_size=10,          # beam search 大小
    lm_weight=0.0,         # 如果沒有語言模型，設置為 0
    word_score=0.0         # 如果沒有語言模型，設置為 0
)

def ctc_decode(features, blank=0):
  if isinstance(features, list):
    features = features[-1]  # 如果是列表，取最後一層作為特徵
  linear = torch.nn.Linear(768, 28)  # vocab_size = 28 (包含 <blank>)
  logits = linear(features)
  probs = F.softmax(logits, dim=-1)  # 使用 softmax 轉換為概率分布
  #print(f"Probs shape: {probs.shape}")  # 應為 [batch_size, sequence_length, vocab_size]
  #print(f"Tokens: {tokens}")  # 確認是否為字符列表
  result, _ = decoder(probs)
  return result

# 設定路徑
wav_root = 'Final_Target_Set/wav/'
decode_root = 'Final_Target_Set/decode/'

# 取得所有 WAV 檔案路徑
wavs_path = [path for path in glob.glob(wav_root + '/*.wav')]

# 檢查並創建解碼輸出資料夾
if not os.path.exists(decode_root):
  os.mkdir(decode_root)

# 載入預訓練模型
model = S3PRLUpstream("wav2vec2")
model.eval()

# 遍歷 WAV 檔案並進行解碼
for wav_path in tqdm(wavs_path, desc="Transcribing"):
  wav_name = wav_path.split('/')[-1].split('.')[0]

  # 讀取音頻檔案
  waveform, sr = librosa.load(wav_path)
  waveform = torch.tensor(waveform).unsqueeze(0)  # 加上 batch 維度

  # 設置音頻長度
  wav_len = torch.LongTensor([waveform.shape[-1]])

  # 提取語音特徵
  with torch.no_grad():
    hidden_states, lengths = model(waveform, wav_len)
  #print(type(hidden_states))
  #for i, tensor in enumerate(hidden_states):
  #  print(f"Tensor {i}: shape={tensor.shape}")

  result = ctc_decode(hidden_states)
  # 要注意result 裡面有沒有 result["text"]

  # 儲存解碼結果
  with open(decode_root + '/' + wav_name + '.txt', 'w') as f:
    f.write(clean_txt(result["text"]) + '\n') # 如果result裡面沒有result["text"]，就不能clean_txt
```

## Eval WER

```python
import jiwer
import glob
# check align
ground_truths_root = 'Final_Target_Set/transcript/'
groundtruths_paths = sorted([path for path in glob.glob(ground_truths_root+'/*.txt')])
hypothesis_paths = sorted([path for path in glob.glob(decode_root+'/*.txt')])
for ele in groundtruths_paths:
  assert ele.split('/')[-1].split('.')[0] == hypothesis_paths[groundtruths_paths.index(ele)].split('/')[-1].split('.')[0]
# load decode result & GT
groundtruths = []
hypothesis = []
for path in groundtruths_paths:
  with open(path,'r') as f:
    groundtruths.append(f.read().replace('\n',''))
for path in hypothesis_paths:
  with open(path,'r') as f:
    hypothesis.append(f.read()[1:].replace('\n',''))
```

```python
wer_score = jiwer.wer(truth=groundtruths, hypothesis=hypothesis)
print(f"{wer_score=}")
```
