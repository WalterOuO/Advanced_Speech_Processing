```python
import os
import librosa
import numpy as np
import soundfile as sf

!mkdir p360
!mv 'Walter_中華隊冠軍_小聲.wav' 'p360/Walter_中華隊冠軍_小聲.wav'
!mv 'Walter_中華隊冠軍啦_大聲.wav' 'p360/Walter_中華隊冠軍啦_大聲.wav'
!mv 'Walter_please_call_Stella.wav' 'p360/Walter_please_call_Stella.wav'

# Walter加入訓練
walter_path = './p360'
files = sorted([f for f in os.listdir(walter_path) if f.endswith('.wav')])  # 確保處理順序

for i, file in enumerate(files):
    full_file_path = os.path.join(walter_path, file)

    # 1. 加速處理
    y, sr = librosa.load(full_file_path, sr=None)  # 保持原取樣率
    # y_speed_up = librosa.effects.time_stretch(y, rate=1.5)  # 加速 1.5 倍

    # 2. 轉換為 16k 取樣率
    y_16k = librosa.resample(y, orig_sr=sr, target_sr=16000)

    # 3. 重命名並儲存
    new_name = f"p360_{i+1:03d}.wav"  # 格式化命名 p360_001.wav, p360_002.wav...
    renamed_path = os.path.join(walter_path, new_name)
    sf.write(renamed_path, y_16k, 16000)  # 儲存加速且轉換為 16k 的音檔

    print(f"Processed and renamed: {file} -> {new_name}")
    # print(f"已加速 1.5 倍並轉換為 16k: {renamed_path}")
```

```python
# 調頻率

def get_main_frequency(file_path):
    """
    計算音檔的主要頻率
    """
    y, sr = librosa.load(file_path, sr=None)
    D = np.abs(librosa.stft(y))
    frequencies = librosa.fft_frequencies(sr=sr)
    mean_spectrum = np.mean(D, axis=1)
    main_frequency = frequencies[np.argmax(mean_spectrum)]
    return main_frequency, sr

# 計算 file1 的主要頻率
file1_path = "p001/p001_468.wav"
file1_main_frequency, sr1 = get_main_frequency(file1_path)
print(f"日本妹 主要頻率: {file1_main_frequency:.2f} Hz")
print()
!mkdir DUMMY/p360
i = 1
for file in os.listdir(walter_path):
  if file.endswith(".wav"):
    if file.startswith("p360_"):
      file_path = os.path.join(walter_path, file)
      file2_main_frequency, sr = get_main_frequency(file_path)
      print(f"中華隊萬歲 主要頻率: {file2_main_frequency:.2f} Hz")
      frequency_ratio = file1_main_frequency / file2_main_frequency
      print(f"頻率比值 (調整倍率): {frequency_ratio:.2f}")
      print()
      y2, sr2 = librosa.load(file_path, sr=None)

      # `n_steps` 計算為頻率比值的對數
      n_steps = 12 * np.log2(frequency_ratio)  # 半音數
      y2_shifted = librosa.effects.pitch_shift(y2, sr=sr2, n_steps=n_steps)

      output_path = f"DUMMY/p360/p360_{i:03d}.wav"
      sf.write(output_path, y2_shifted, sr2)
      i += 1
```
